## 9) Código funcional


```python
class Pila:
    def __init__(self):
        self.items = []

    def push(self, elemento):
        self.items.append(elemento)

    def pop(self):
        if not self.esta_vacia():
            return self.items.pop()

    def top(self):
        if not self.esta_vacia():
            return self.items[-1]

    def esta_vacia(self):
        return len(self.items) == 0

    def tamano(self):
        return len(self.items)


class Cola:
    def __init__(self):
        self.items = []

    def encolar(self, elemento):
        self.items.append(elemento)

    def desencolar(self):
        if not self.esta_vacia():
            return self.items.pop(0)

    def frente(self):
        if not self.esta_vacia():
            return self.items[0]

    def esta_vacia(self):
        return len(self.items) == 0

    def tamano(self):
        return len(self.items)


def sistema_medico():

    n = int(input())

    cola_normales = Cola()
    cola_criticos = Cola()

    for _ in range(n):

        accion = input().split()

        if accion[0] == "N":

            nombre = accion[1]
            cola_normales.encolar(nombre)
            print("Paciente N nuevo:", nombre)

        elif accion[0] == "C":

            nombre = accion[1]
            cola_criticos.encolar(nombre)
            print("Paciente C nuevo:", nombre)

        elif accion[0] == "A":

            if not cola_criticos.esta_vacia():

                paciente = cola_criticos.desencolar()
                print("Se atiende a", paciente)

            elif not cola_normales.esta_vacia():

                paciente = cola_normales.desencolar()
                print("Se atiende a", paciente)

            else:
                print("no hay pacientes")


sistema_medico()
```
