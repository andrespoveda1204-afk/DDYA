# Queue using two stacks


## Enunciado

El enunciado del problema basicamente dice que, debemos crear una cola con 2 pilas, y a base de esa cola, tendremos un numero entero positivo el cual determinara el numero de consultas para 3 opciones que nos dan con la cola:
- 1. Encolar un elemento al final de la cola
- 2. Desencolar el elemento que este al frente de la cola ( es decir el primero ingresado)
- 3. Imprimir el elemento que esta al frente de la cola


##Especificación

El primer paso para desarrolar este problema seria, la creacion de la cola utilizando dos pilas, eso dentro del sistema, ahora.
- **Precondicion (Posibles entradas):**
 con entrada en el numero de consultas, se espera un numero entero positivo n > 0, una vez cumplida esra precondicion, ahora pasamos con los 3 pasos posibles vistos en el enunciado las n veces de consultas, aqui las posibles entradas son: 1, 2, 3.    Nada diferente a eso en las n consultas, si es cualquier entrada diferente, el sistema mostrara un mensaje de correcion para la entrada que se espera, si se elige la opcion 1, sera la de encolar algun elemento, aqui la entrada no tiene restricciones, el usuario puede poner en esta opcion lo que quiera, para la opcion 2 no se tiene entrada, simplemente es un paso dentro del sistema, e igual para la eleccion del paso 3.
 ----------
 - **Poscondicion (salidas esperadas):**
Primeramente, con el ingreso del numero n de consultas, obviamente si esta dentro de las posibles entradas, se imprimira el menu con las posibles opciones, despues, si se elige la opcion 1, la salida esperada es un mensaje preguntando el valor a encolar, en la opcion 2 no tendremos una salida esperada y con la opcion 3, aqui la salida que se espera, es imprimir el primer valor que se agrego con la opcion 1 (si se ha eliminado con la opcion 2, pues el seiguiente que se agrego, asi consecutivamente), siguiendo la regla de las colas, FIFO, y si tenemos nuestra cola esta vacia la salida esperada es nada, se devuelve al menu de opciones si quedan pasos.

## Análisis de complejidad
para determinar la complejidad de mi solucion al problema, vamos a analizar el codigo en 2 partes, las funciones de mi clase donde realizaremos las operaciones dequeue, enqueue, y peek(impresion del elemento en frente) todas estas operaciones individuales son O(1), pero en esta parte del codigo tenemos otra funcion, que seria el peor caso de esta parte, la funcion preparar_salida que esta funcion se ejecuta cuando la pila.salida esta vacia, y lo que hace esta funcion es un bucle que pasa todos los elementos de la pila.salida a la pila.salida, esta operacion es O(n).
##Class ColaconPilas:
- **Mejor caso:** O(1)
- **peor caso:** O(n)

Ahora vamos a analizar la complejidad de mi otra parte del codigo, lo que tengo dentro del main, aqui se ejecuta un bucle con el numero n de consultas, asi que seria una complejidad temporal O(n), pero, mi codigo es eficiente, ya que dentro del bucle, se llaman las funciones de mi clase ColaconPilas para realizar las 3 diferentes opciones y como visto anteriormente, estas operaciones son O(1) asi que la complejidad total de esta parte es O(n).
## Main:
- **Mejor caso:** O(n)
- **Peor caso:** O(n)

**Complejidad total**:

O(n) + O(n) = O(n)



## Codigo
*En esta sección se escribe el código de la solución que se ha diseñado para resolver el problema. *


```python
#Código
class ColaConPilas:
    def __init__(self):
        self.pila_entrada = []
        self.pila_salida = []

    def enqueue(self, x):
        # Simplemente agregamos a la pila de entrada
        self.pila_entrada.append(x)

    def preparar_salida(self):
        # Si la pila de salida está vacía, movemos todo de la entrada a la salida
        if not self.pila_salida:
            while self.pila_entrada:
                item = self.pila_entrada.pop()
                self.pila_salida.append(item)

    def dequeue(self):
        self.preparar_salida()
        if self.pila_salida:
            return self.pila_salida.pop()

    def peek(self):
        self.preparar_salida()
        if self.pila_salida:
            return self.pila_salida[-1] # Mirar el de arriba sin sacarlo


def main ():
    mi_cola = ColaConPilas()
    print("\nSISTEMA EJECUTADO CON EXITO")
    print("---------------------------")
    q = int(input("Ingresa la cantidad de consultas que deseas: "))
    cont = 0
    while cont < q:
    # Bucle infinito de validación
        while True:
         try:
            print("\n--- MENÚ DE OPCIONES ---")
            print("1. Enqueue (Agregar)")
            print("2. Dequeue (Eliminar)")
            print("3. Print (Mostrar frente)")
            opcion = int(input("\nElige una opcion: "))
            if 1 <= opcion <= 3:
                if opcion == 1:
                  x = input("ingresa el elemento a encolar: ")
                  mi_cola.enqueue(x)           # Pides el valor 'x' y haces enqueue [cite: 29]
                  pass
                elif opcion == 2:
                  mi_cola.dequeue()
                  pass
                elif opcion == 3:
                  frente = mi_cola.peek()
                  if frente is not None: #Significa que si la variable frente no esta vacia, asi que como no lo esta imprimimos lo que tiene
                   print (frente) # Sale del bucle de validación si 3es correcto
                break
            else:
                print("\nError: Ingresa un paso permitido (1-3)")
         except ValueError:
            print("Error: Debes ingresar un número entero")
        cont +=1
main ()
```

```python
class ColaConPilas:
    def __init__(self):
        self.pila_entrada = []
        self.pila_salida = []

    def enqueue(self, x):
        # Simplemente agregamos a la pila de entrada
        self.pila_entrada.append(x)

    def preparar_salida(self):
        # Si la pila de salida está vacía, movemos todo de la entrada a la salida
        if not self.pila_salida:
            while self.pila_entrada:
                item = self.pila_entrada.pop()
                self.pila_salida.append(item)

    def dequeue(self):
        self.preparar_salida()
        if self.pila_salida:
            return self.pila_salida.pop()

    def peek(self):
        self.preparar_salida()
        if self.pila_salida:
            return self.pila_salida[-1] # Mirar el de arriba sin sacarlo
def bench(n: int, step: int = 1) -> list[tuple[int, float]]:
  """
   Mide el tiempo de ejecución de una función para distintos tamaños de entrada.

   Parameters:
    - n: tamaño máximo de entrada.
    - step: incremento entre pruebas (por defecto 1).

   Return:
    - Lista de tuplas (tamaño, tiempo en segundos).
  """

  print("Generando datos ...", end='')
  from random import randint # para generar numeros aleatorios
  import time
  results: list[tuple[int, float]] = []
  for size in range(1, n, step):
    # 1. PREPARACIÓN (Esto no se mide)
    # Creamos una cola y la llenamos con 'size' elementos
    test_cola = ColaConPilas()
    for i in range(size):
        test_cola.enqueue(i)

    # 2. MEDICIÓN (Calculamos cuánto tarda en sacar el frente)
    ti = time.time()

    # Aquí ejecutamos la operación que queremos probar (ej: dequeue o peek)
    test_cola.dequeue()

    tf = time.time()

    # 3. REGISTRO
    delta = tf - ti
    print(".", end="")
    results.append((size, delta))
  print("Finalizado")
  return results

def dibujar(x, y, dimensiones=(7, 5)):
  """
  Dibuja una serie de puntos en un plano cartesiano
  Parameters:
  - x: lista de numeros
  - y: lista de numeros
  """
  assert len(x) == len(y), "la cantidad de numeros en ambas debe ser el mismo"

  import matplotlib.pyplot as plt
  plt.figure(figsize=dimensiones)
  plt.plot(x, y)
  plt.show()

```

```python
tiempos = bench(10_000_000, step = 250_000)

# Los convertimos a coordenadas para dibujar una grafica
x = [size for size, timing in tiempos]
y = [timing for size, timing in tiempos]

dibujar(x, y)
```
